###  3. Ambil contoh threshold split (misalnya):

* Threshold = 265 cm

Maka kita akan membagi:

* Group Left (≤ 265)
* Group Right (> 265)

###  4. Hitung Entropy Awal (sebelum split)

$$
Entropy(D) = -p_1 \log_2(p_1) - p_2 \log_2(p_2)
$$

$$
p_1 = \frac{8}{15},\quad p_2 = \frac{7}{15}
$$

$$
Entropy(D) = -\frac{8}{15} \log_2 \frac{8}{15} - \frac{7}{15} \log_2 \frac{7}{15} \approx 0.996
$$

###  5. Entropy setelah split

Group Left (≤ 265):

* Data: No 13, 9, 6, 15, 8, 7, 4, 10 → 8 data
* Kelas: 7 Panthera leo, 1 Panthera tigris

$$
Entropy(left) = -\frac{7}{8}\log_2\frac{7}{8} - \frac{1}{8}\log_2\frac{1}{8} \approx 0.543
$$

Group Right (> 265):

* Data: No 12, 3, 1, 14, 2, 11, 5 → 7 data
* Kelas: 7 Panthera tigris

$$
Entropy(right) = -\frac{7}{7} \log_2 \frac{7}{7} = 0
$$

Information Gain:

$$
IG = Entropy(D) - \left( \frac{8}{15} \cdot 0.543 + \frac{7}{15} \cdot 0 \right) \approx 0.996 - 0.2896 \approx \boxed{0.7064}
$$

###  6. Gini Index

$$
Gini(D) = 1 - \sum p_i^2 = 1 - \left(\frac{8}{15}\right)^2 - \left(\frac{7}{15}\right)^2 \approx 0.497
$$

Group Left (≤ 265):

* 7 Panthera leo, 1 Panthera tigris

$$
Gini(left) = 1 - \left(\frac{7}{8}\right)^2 - \left(\frac{1}{8}\right)^2 = 1 - 0.765625 - 0.015625 = 0.21875
$$

Group Right (> 265):

* 7 Panthera tigris

$$
Gini(left) = 1 - \left(\frac{7}{7}\right)^2 = 1 - 1 = 0
$$

Gini Split:

$$
Gini_{split} = \frac{8}{15} \cdot 0.21875 + \frac{7}{15} \cdot 0 = \boxed{0.116}
$$

###  2. Ambil threshold split (contoh):

Gunakan threshold = 200 kg (terjadi perpindahan mayoritas dari *leo* ke *tigris*)

#### Maka:

* Group Left (≤ 200) = data dengan berat ≤ 200 kg (No: 13, 9, 6, 15, 8, 7, 4, 10) → 8 data
* Group Right (> 200) = sisanya (No: 2, 14, 5, 1, 12, 3, 11) → 7 data


###  3. Komposisi kelas:

Sebelum split:

* 8 *Panthera tigris*
* 7 *Panthera leo*

$$
Entropy(D) = -\frac{8}{15} \log_2 \frac{8}{15} - \frac{7}{15} \log_2 \frac{7}{15} \approx 0.996
$$


###  4. Entropy setelah split:

#### Group Left (≤ 200 kg):

* 6 *Panthera leo*: No 13, 9, 6, 15, 8, 7
* 2 *Panthera tigris*: No 4, 10

$$
Entropy(left) = -\frac{6}{8} \log_2 \frac{6}{8} - \frac{2}{8} \log_2 \frac{2}{8} \\
= -0.75 \log_2 0.75 - 0.25 \log_2 0.25 \\
\approx -0.75(-0.415) - 0.25(-2) \\
\approx 0.311 + 0.5 = \boxed{0.811}
$$

#### Group Right (> 200 kg):

* 7 *Panthera tigris*: No 2, 14, 5, 1, 12, 3, 11

$$
Entropy(right) = -\frac{7}{7} \log_2 \frac{7}{7} = 0
$$


###  5. Information Gain

$$
IG = Entropy(D) - \left( \frac{8}{15} \cdot 0.811 + \frac{7}{15} \cdot 0 \right) \\
= 0.996 - 0.432 = \boxed{0.564}
$$


###  6. Gini Index

$$
Gini(D) = 1 - \left(\frac{8}{15}\right)^2 - \left(\frac{7}{15}\right)^2 \approx 0.497
$$

#### Gini(Left):

$$
Gini = 1 - (6/8)^2 - (2/8)^2 = 1 - 0.5625 - 0.0625 = \boxed{0.375}
$$

#### Gini(Right):

Semua satu kelas → Gini = 0

#### Gini Split:

$$
Gini_{split} = \frac{8}{15} \cdot 0.375 + \frac{7}{15} \cdot 0 = \boxed{0.2}
$$

## Visualisasi Decision Tree

Ini adalah code program untuk mencari decision tree sekaligus rules

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.preprocessing import LabelEncoder

# === Data ===
data = {
    "Panjang Tubuh (cm)": [280, 290, 275, 265, 285, 250, 260, 255, 245, 265, 295, 270, 240, 280, 250],
    "Berat Tubuh (kg)": [220, 210, 230, 190, 215, 180, 190, 185, 175, 200, 240, 220, 170, 210, 185],
    "Kelas": [
        "Panthera tigris", "Panthera tigris", "Panthera tigris", "Panthera tigris", "Panthera tigris",
        "Panthera leo", "Panthera leo", "Panthera leo", "Panthera leo", "Panthera leo",
        "Panthera tigris", "Panthera tigris", "Panthera leo", "Panthera tigris", "Panthera leo"
    ]
}
df = pd.DataFrame(data)

# === Label Encoding ===
le = LabelEncoder()
df['Label'] = le.fit_transform(df['Kelas'])  # tigris=1, leo=0

# === Fitur dan Target ===
X = df[["Panjang Tubuh (cm)", "Berat Tubuh (kg)"]]
y = df["Label"]  # <- gunakan label numerik hasil encoding

# === Train Decision Tree ===
clf = DecisionTreeClassifier(criterion="gini", max_depth=3, random_state=42)
clf.fit(X, y)

# === 1. Tampilkan Pohon Keputusan ===
print("\n=== Pohon Keputusan (Decision Tree) ===")
tree_rules = export_text(clf, feature_names=list(X.columns))
print(tree_rules)
```

## Visualisasi Splitting

```python
# === Uji 2 Data Baru ===
uji_baru = pd.DataFrame({
    'Panjang Tubuh (cm)': [235, 248],
    'Berat Tubuh (kg)': [225, 215]
})

prediksi = clf.predict(uji_baru)
hasil = le.inverse_transform(prediksi)

print("\n=== Prediksi 2 Data Baru ===")
for i, (p, b) in enumerate(zip(uji_baru['Panjang Tubuh (cm)'], uji_baru['Berat Tubuh (kg)'])):
    print(f"Data {i+1}: Panjang={p}, Berat={b} → Prediksi: {hasil[i]}")

# === Visualisasi Decision Boundary ===
print("\n=== Visualisasi Decision Boundary ===")
plot_colors = "ry"
plot_step = 1.0

x_min, x_max = X['Panjang Tubuh (cm)'].min() - 10, X['Panjang Tubuh (cm)'].max() + 10
y_min, y_max = X['Berat Tubuh (kg)'].min() - 10, X['Berat Tubuh (kg)'].max() + 10

xx, yy = np.meshgrid(np.arange(x_min, x_max, plot_step),
                     np.arange(y_min, y_max, plot_step))

grid = np.c_[xx.ravel(), yy.ravel()]
grid_df = pd.DataFrame(grid, columns=['Panjang Tubuh (cm)', 'Berat Tubuh (kg)'])
Z = clf.predict(grid_df)
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, cmap=plt.cm.RdYlBu, alpha=0.6)

# Plot titik data training
for label, color in zip([0, 1], plot_colors):
    idx = np.where(y == label)
    plt.scatter(X.iloc[idx]['Panjang Tubuh (cm)'], X.iloc[idx]['Berat Tubuh (kg)'],
                c=color, edgecolor='k', label=le.inverse_transform([label])[0])

plt.xlabel("Panjang Tubuh (cm)")
plt.ylabel("Berat Tubuh (kg)")
plt.title("Decision Boundary (Panjang vs Berat)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# === Visualisasi Decision Boundary ===
print("\n=== Visualisasi Decision Boundary ===")
plot_colors = "ry"
plot_step = 1.0

x_min, x_max = X['Panjang Tubuh (cm)'].min() - 10, X['Panjang Tubuh (cm)'].max() + 10
y_min, y_max = X['Berat Tubuh (kg)'].min() - 10, X['Berat Tubuh (kg)'].max() + 10

xx, yy = np.meshgrid(np.arange(x_min, x_max, plot_step),
                     np.arange(y_min, y_max, plot_step))

grid = np.c_[xx.ravel(), yy.ravel()]
grid_df = pd.DataFrame(grid, columns=['Panjang Tubuh (cm)', 'Berat Tubuh (kg)'])
Z = clf.predict(grid_df)
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, cmap=plt.cm.RdYlBu, alpha=0.6)

# Plot titik data training
for label, color in zip([0, 1], plot_colors):
    idx = np.where(y == label)
    plt.scatter(X.iloc[idx]['Panjang Tubuh (cm)'], X.iloc[idx]['Berat Tubuh (kg)'],
                c=color, edgecolor='k', label=le.inverse_transform([label])[0])

# Plot data uji
plt.scatter(uji_baru['Panjang Tubuh (cm)'], uji_baru['Berat Tubuh (kg)'],
            c='black', marker='X', s=100, label='Data Uji')

plt.xlabel("Panjang Tubuh (cm)")
plt.ylabel("Berat Tubuh (kg)")
plt.title("Decision Boundary (Panjang vs Berat)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Visualisasi Decision Tree dalam bentuk gambar

```python
# === Visualisasi Struktur Pohon ===
plt.figure(figsize=(12, 3))
plot_tree(
    clf,
    feature_names=["Panjang Tubuh (cm)", "Berat Tubuh (kg)"],
    class_names=le.classes_,  # gunakan label string hasil encoding
    filled=True,
    rounded=True,
    impurity=True  # tampilkan nilai Gini
)
plt.title("Decision Tree untuk Panthera tigris vs Panthera leo")
plt.tight_layout()
plt.show()
```
